```python
import yfinance as yf
import pandas as pd

# List of stock tickers
tickers = ["III.L", "AAF.L"]

all_data = pd.DataFrame()

for i in tickers:
    data = yf.download(i, period = "1y", interval = '1d')
    data = data.reset_index()
    data["Ticker"] = i
    data = data[['Date', 'Close', 'High', 'Low', 'Open', 'Volume', 'Ticker']]
    all_data = pd.concat([all_data, data])

# all_data.reset_index=True
all_data.head()
```

```python
import yfinance as yf
import pandas as pd

tickers = ["III.L", "AAF.L", "ANTO.L", "AZN.L", "BME.L", "BAB.L",
           "BA.L", "BARC.L", "BWY.L", "BKG.L", "BP.L", "BRBY.L"]

all_data = pd.DataFrame()

for ticker in tickers:
    data = yf.download(ticker, period="1y", interval="1d")
    
    if isinstance(data.columns, pd.MultiIndex):
        data.columns = data.columns.get_level_values(0)

    data = data.reset_index()
    data['Ticker'] = ticker
    data = data[['Date', 'Open', 'High', 'Low', 'Close', 'Volume', 'Ticker']]
    all_data = pd.concat([all_data, data], ignore_index=True)

```
